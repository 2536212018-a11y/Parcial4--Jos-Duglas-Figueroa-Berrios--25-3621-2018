```python
# =========================================
# PREPARACION DE DATOS PARA APRIORI
# =========================================

# Agrupar productos por transaccion

transacciones = asociacion.groupby(['transaccion_id', 'item'])['cantidad'].sum().unstack().fillna(0)

# Mostrar primeras filas

transacciones.head()
```

```python
# Convertir cantidades a valores binarios

def encode(x):
    if x > 0:
        return 1
    else:
        return 0

transacciones_binarias = transacciones.applymap(encode)

# Mostrar resultado

transacciones_binarias.head()
```

```python
# Instalar libreria para Apriori

!pip install mlxtend
```

```python
# Importar algoritmo Apriori

from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
```

```python
# =========================================
# APLICAR ALGORITMO APRIORI
# =========================================

# Generar combinaciones frecuentes

frequent_itemsets = apriori(
    transacciones_binarias,
    min_support=0.02,
    use_colnames=True
)

# Mostrar resultados

frequent_itemsets.head(10)
```

```python
# =========================================
# GENERAR REGLAS DE ASOCIACION
# =========================================

# Crear reglas de asociacion

reglas = association_rules(
    frequent_itemsets,
    metric="lift",
    min_threshold=1
)

# Mostrar primeras reglas

reglas.head()
```

```python
# Mostrar columnas importantes

reglas_filtradas = reglas[[
    'antecedents',
    'consequents',
    'support',
    'confidence',
    'lift'
]]

reglas_filtradas.head(10)
```

```python
# Ordenar reglas por lift

top_reglas = reglas_filtradas.sort_values(by='lift', ascending=False)

top_reglas.head(10)
```

```python
# =========================================
# EJERCICIO 3 - CLUSTERING
# =========================================

# Mostrar primeras filas

agrupacion.head()
```

```python
# =========================================
# SELECCION DE VARIABLES
# =========================================

# Variables seleccionadas

variables_cluster = agrupacion[[
    'ingresos',
    'frecuencia_uso',
    'gasto_promedio',
    'satisfaccion',
    'reclamos'
]]

# Mostrar variables

variables_cluster.head()
```

```python
# =========================================
# ESCALADO DE VARIABLES
# =========================================

from sklearn.preprocessing import StandardScaler
```

```python
# Crear escalador

scaler = StandardScaler()

# Escalar variables

variables_escaladas = scaler.fit_transform(variables_cluster)

# Mostrar resultado

variables_escaladas[:5]
```

```python
# =========================================
# METODO DEL CODO
# =========================================

from sklearn.cluster import KMeans
```

```python
# Revisar valores nulos

variables_cluster.isnull().sum()
```

```python
# =========================================
# LIMPIAR VALORES NULOS
# =========================================

# Reemplazar nulos con la media

variables_cluster = variables_cluster.fillna(
    variables_cluster.mean()
)

# Verificar nuevamente

variables_cluster.isnull().sum()
```

```python
# Escalar nuevamente los datos

variables_escaladas = scaler.fit_transform(variables_cluster)
```

```python
inercias = []

for k in range(1, 11):

    modelo = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    modelo.fit(variables_escaladas)

    inercias.append(modelo.inertia_)
```

```python
# Graficar metodo del codo

plt.figure(figsize=(8,5))

plt.plot(range(1,11), inercias, marker='o')

plt.xlabel('Numero de clusters')

plt.ylabel('Inercia')

plt.title('Metodo del Codo')

plt.show()
```

```python
# =========================================
# METODO DEL CODO
# =========================================

inercias = []

for k in range(1, 11):

    modelo = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    modelo.fit(variables_escaladas)

    inercias.append(modelo.inertia_)

inercias
```

```python
plt.figure(figsize=(8,5))

plt.plot(range(1,11), inercias, marker='o')

plt.xlabel('Numero de clusters')
plt.ylabel('Inercia')
plt.title('Metodo del Codo')

plt.show()
```

```python
# =========================================
# APLICAR K-MEANS
# =========================================

# Crear modelo K-Means

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

# Entrenar modelo

kmeans.fit(variables_escaladas)

# Mostrar clusters encontrados

kmeans.labels_
```

```python
# =========================================
# AGREGAR CLUSTERS AL DATASET
# =========================================

# Agregar columna cluster

agrupacion['cluster'] = kmeans.labels_

# Mostrar resultado

agrupacion.head()
```

```python
# =========================================
# VISUALIZACION DE CLUSTERS
# =========================================

plt.figure(figsize=(10,6))

# Scatter plot

plt.scatter(
    agrupacion['ingresos'],
    agrupacion['gasto_promedio'],
    c=agrupacion['cluster'],
    cmap='viridis'
)

# Etiquetas

plt.xlabel('Ingresos')

plt.ylabel('Gasto Promedio')

plt.title('Clusters de Clientes')

# Mostrar grafico

plt.show()
```
